```python
import os
from swed_17.snow17 import MADLY

import matplotlib.pyplot as plt
```

```python
db_query = MADLY(os.environ.get("S17_DB"))
```

```python
zone = "ALTC2LUF"
ua_swe = db_query.to_df(zone, MADLY.UofA, 2023)
snodas_swe = db_query.to_df(zone, MADLY.SNODAS, 2023)
```

```python
fig = plt.figure(dpi=200)
ax = fig.gca()
ua_swe["SWE (mm)"].plot(ax=ax, label=MADLY.UofA)
snodas_swe["SWE (mm)"].plot(ax=ax, label=MADLY.SNODAS)
ax.legend()
```
